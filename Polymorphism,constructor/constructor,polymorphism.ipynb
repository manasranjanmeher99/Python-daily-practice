```python
class A():
    def __init__(self):
        print("A init")
    def feature1(self):
        print("feature1 is hi")
    def feature2(self):
        print("feature2 is hi")

class B(A):
    def __init__(self):
        super().__init__()  # This will call the constructor of class A
        print("B init")
        
    def feature3(self):
        print("feature3 is hello")
    def feature4(self):
        print("feature4 is hello")


b1= B() # print("A init") will be printed first because we have called the constructor of class A using super().__init__() in the constructor of class B. Then, print("B init") will be printed.
```

```python
class A():
    def __init__(self):
        print("A init")
    def feature1(self):
        print("feature1 is hi")
    def feature2(self):
        print("feature2 is hi")

class B():
    def __init__(self):
        print("B init")
    def feature3(self):
        print("feature3 is hello")
    def feature4(self):
        print("feature4 is hello")

class C(B, A):
    def __init__(self):
        super().__init__()  # This will call the constructor of class B because class B is the first parent class in the inheritance list.
        print("C init")

c1= C()
# print("B init") will be printed first because we have called the constructor of class B using super().__init__() in the constructor of class C. 
# Then, print("C init") will be printed.
# Beacause class B is the first parent class in the inheritance list of class C, the constructor of class B will be called first.
```

```python
class student():
    def __init__(self,m1,m2):
        self.m1 = m1
        self.m2 = m2
        
    def __add__(self, other): ## WE DEFINE THE ADD METHOD . THIS CONCEPT IS CALLED OPERATOR OVERLOADING
        m1 = self.m1 + other.m1
        m2 = self.m2 + other.m2
        s3 = student(m1,m2)
        
        return s3
    def __gt__(self,other):
        r1 = self.m1 + self.m2
        r2 = other.m1 + other.m2
        if r1 > r2:
            return True
        else:
            return False   
 
s1 = student(60,70)
s2 = student(80,90)

s3 = s1 + s2 
#we can also compare 2 object are > or less then 
if s1>s2: 
    print('s1 pass')
else:
    print('s2 pass')
```

# Method overloading

```python
class demo():
    def add(self,a,b):
        return a+b

d = demo()
print(d.add(2,3))
```

```python
class demo():
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c
    def add(self,a,b,c,d):
        return a+b+c+d

d= demo()
print(d.add(2,3))
print(d.add(2,3,4))
print(d.add(2,3,4,5))
```

```python
class demo():
    def add(self,a,b,c=0, d=0):
        return a+b+c+d
    
d= demo()
print(d.add(2,3))
print(d.add(2,3,4))
print(d.add(2,3,4,5))
```

```python
class demo():
    def add(self, *args):
        total=0
        for i in args:
            total = total + i
        return total

d= demo()
print(d.add(2,3))
print(d.add(2,5,4))
print(d.add(2,3,4,5,6,7,8,9))
```
