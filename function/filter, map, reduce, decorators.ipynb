# 1. filter() Function

- filter() is used to filter elements from a list based on a condition.

  filter(function, iterable)


```python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

even= list(filter(lambda x: x%2==0, numbers))
print(even)
```

```python
names = ["Alice", "Bob", "Ankit", "Rahul"]

result = list(filter(lambda x: x.startswith("A"),names))
print(result)
```

# 2. map() function

- map() is used to transform each element in a list.

    map(function, iterable)

```python
numbers= [1, 2, 3, 4, 5]

squares= list(map(lambda x: x*x, numbers))
print(squares)
```

```python
numbers= [1, 2, 3, 4, 5]

squares= list(map(lambda x: x**2, numbers))
print(squares)
```

```python
names = ["python", "java", "sql", "javascript"]

uppercase= list(map(lambda x: x.upper(), names))
print(uppercase)
```

```python
names = ['PYTHON', 'JAVA', 'SQL', 'JAVASCRIPT']

lowercase= list(map(lambda x: x.lower(), names))
print(lowercase)
```

# 3. reduce() function:

- reduce() is used to reduce all elements into a single value.

- It is available in the functools module.

        from functools import reduce

        reduce(function, iterable)

```python
# calculate the total sum

from functools import reduce
num = [3,2,6,8,4,6,2]
total = reduce(lambda a,b : a+b, num)
print("Total sum =",total)
```

```python
# find maximum number

num = [3,2,6,8,4,6,2]
maximum= reduce(lambda a,b: a if a>b else b, num)
print("maximum =",maximum)
```

```python
# find minimum value

num= [3,2,6,8,4,6,9,1]
minimum= reduce(lambda a,b: a if a<b else b, num)
print("minimum =",minimum)
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5, 6]

# Step 1: Filter even numbers
even = list(filter(lambda x: x % 2 == 0, numbers))

# Step 2: Square them
square = list(map(lambda x: x ** 2, even))

# Step 3: Sum all values
add = reduce(lambda x, y: x + y, square)

print("Even: ",even)
print("Square: ",square)
print("Addition: ",add)
```

# Python Decorators

- A decorator in Python is used to modify or extend the behavior of a function without changing its original code.
- A function can be passed to another function
- A function can return another function
- Python provides a shortcut syntax using @.

```python
def decorator(func):
    def wrapper():
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper

@decorator
def greet():
    print("Hello, World!")
greet()
```
