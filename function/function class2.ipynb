# factorial

```python
def fact(n):
    if n==0:
        return 1
    return n* fact(n-1)

result= fact(5)
print("Factorial is: ", result)
```

# Febonacci series

```python
def febo(n):
    a=0
    b=1
    print("Febonaci series:")
    print(a)
    print(b)
    for i in range(1,n-1):
        c=a+b
        a=b
        b=c
        print(c)

result= febo(10)
```

```python
# Simple recursive function that counts down from 5

def countdown(n):
    if n<=0:
        return 1
    else:
        print(n)
        countdown(n-1)
countdown(5)
```

```python
# Fibonacci sequence using recursion function

def fibonacci(n):
  if n <= 1:
    return n
  else:
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(7))
```

```python
# Calculate the sum of all elements in a list:

def sum_list(numbers):
  if len(numbers) == 0:
    return 0
  else:
    return numbers[0] + sum_list(numbers[1:])

my_list = [1, 2, 3, 4, 5]
print(sum_list(my_list))
```
